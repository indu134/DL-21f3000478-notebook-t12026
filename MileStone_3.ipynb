```python
#####################################################################
#
import torch
import torch.nn as nn
import torchaudio
import numpy as np
```

```python
#######################################################################
#Question-1
# batch of grayscale spectrograms
t = torch.randn(32, 1, 64, 64)

# flatten everything except batch dimension
flattened_t = t.view(32, -1)

print("Original shape :", t.shape)
print("Flattened shape:", flattened_t.shape)
print("Input features for Linear layer:", flattened_t.shape[1])
```

```python
#####################################################################
#Question-2

sample_rate = 16000
audio_data     = torch.randn(1, sample_rate)

# mel spectrogram config
mel_transform = torchaudio.transforms.MelSpectrogram(
    sample_rate = sample_rate,
    n_fft       = 400,
    hop_length  = 160,
    n_mels      = 64
)

mel_data = mel_transform(audio_data)

print("Audio shape      :", audio_data.shape)
print("Mel Spectrogram  :", mel_data.shape)
```

```python
###############################################################################
#Question-3
# input shape (1, 64, 64)
x = torch.randn(1, 1, 64, 64)

# conv layer with padding=1 keeps same size
conv = nn.Conv2d(1, 16, kernel_size=3, stride=1, padding=1)

# maxpool halves the size
pool = nn.MaxPool2d(kernel_size=2, stride=2)

output = pool(conv(x))

print("Input shape  :", x.shape)
print("After Conv   :", conv(x).shape)
print("After MaxPool:", output.shape)
print("Spatial output (H, W):", output.shape[2], output.shape[3])
```

```python
###########################################################################
#Question-4
layer = nn.Linear(in_features=128, out_features=64)

weights = layer.weight.numel()
biases  = layer.bias.numel()
total   = weights + biases

print("Weights    :", weights)
print("Biases     :", biases)
print("Total params:", total)
```

```python
##################################################################################
#Question-5
import math

total_files = 1050
batch_size  = 32

num_batches = math.ceil(total_files / batch_size)

print("Total files :", total_files)
print("Batch size  :", batch_size)
print("Num batches :", num_batches)
```

```python
#################################################################################
#Question-6
logits_i = torch.tensor([[2.5, 1.0, 0.1]])
target_i = torch.tensor([0])

criterion = nn.CrossEntropyLoss()
loss      = criterion(logits_i, target_i)

print("Logits :", logits_i)
print("Target : Class 0")
print("Loss   :", round(loss.item(), 3))
```

```python
####################################################################################
#Question-7
current_weight = 0.5
gradient       = 0.2
learning_rate  = 0.01

new_weight = current_weight - learning_rate * gradient

print("Current weight :", current_weight)
print("Gradient       :", gradient)
print("Learning rate  :", learning_rate)
print("New weight     :", new_weight)
```

```python
######################################################################################
#Question-8
kernel_size = 5
padding     = (kernel_size - 1) // 2

x    = torch.randn(1, 1, 64, 64)
conv = nn.Conv2d(1, 1, kernel_size=5, stride=1, padding=padding)
output  = conv(x)

print("Kernel size    :", kernel_size)
print("Padding needed :", padding)
print("Input shape    :", x.shape)
print("Output shape   :", output.shape)
```
